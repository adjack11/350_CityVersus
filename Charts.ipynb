```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('city_stats.csv')
labels = ['White', 'Black', 'American Indian', 'Asian', 'Two Plus Races', 'Hispanic']
for y in range(0, len(df)):
    current_city = df.loc[y]
    values = [current_city['white_percent'], current_city['black_percent'], current_city['american_indian_percent'], current_city['asian_percent'], current_city['two_plus_races_percent'], current_city['hispanic_percent']]
    x = np.arange(len(labels))
    width = 0.35
    fig, ax = plt.subplots(figsize=(10,6))
    plt.bar(labels, values)
    plt.title('Race Diversity by Percentage')
    plt.xlabel('Race')
    plt.ylabel('Percentage')
    if y == 0:
        plt.savefig('lou_race_diversity.png')
    elif y == 1:
        plt.savefig('cinci_race_diversity.png')
    elif y  == 2:
        plt.savefig('colum_race_diversity.png')
    elif y == 3:
        plt.savefig('clev_race_diversity.png')
    elif y == 4:
        plt.savefig('indy_race_diversity.png')
    elif y == 5:
        plt.savefig('st_race_diversity.png')
    elif y == 6:
        plt.savefig('det_race_diversity.png')
    elif y == 7:
        plt.savefig('mem_race_diversity.png')
    elif y == 8:
        plt.savefig('nash_race_diversity.png')
    else:
        plt.savefig('chi_race_diversity.png')
    plt.show()
    plt.close()
```
